```python
import pandas as pd

link_csv = "https://raw.githubusercontent.com/2005c-h/EstadisticaInferencial2026A/refs/heads/main/Unidad01/Problemario/Dato_problema_01.csv"

df = pd.read_csv(link_csv)

df
```

1. Se eligio estas variables por que con el promedio de los exámenes cortos puede analizar si ayuda a explicar el promedio final

```python
x = df["promedio_de_examenes_cortos"]
y = df["promedio_final"]
```

2. Con el grafico se observa que existe una relación positiva entre ambas variables

```python
import matplotlib.pyplot as plt
import numpy as np

# --- Configuración general del gráfico ---
plt.figure(
    figsize=(6, 4),   # tamaño de la figura (ancho, alto) en pulgadas
    dpi=120           # resolución del gráfico
)

# --- Gráfico de dispersión ---
plt.scatter(
    x, y,
    marker="o",       # forma
    color='blue',     # color de los puntos
    edgecolor='black',    # borde de los puntos
    alpha=0.8,            # transparencia
    s=30,                 # tamaño de los puntos
    label='Datos originales' # etiqueta para la leyenda
)

# --- Calcular la recta de regresión ---
# Ajustar un polinomio de grado 1 (regresión lineal) a los datos x e y
coeficientes = np.polyfit(x, y, 1)
# Crear una función polinómica a partir de los coeficientes
recta_regresion = np.poly1d(coeficientes)
# Calcular los valores y predichos (y_calculada) usando la recta de regresión
y_calculada = recta_regresion(x)

# --- Gráfico de línea ---
plt.plot(
    x, y_calculada,
    color='black',   # color de la línea
    linewidth=1.0,        # grosor de la línea
    linestyle='-',        # estilo de línea (cambiado de '--' a '-')
    marker='',           # no necesitamos marcador en la línea de regresión
    markersize=0,         # tamaño del marcador
    markerfacecolor='white',
    markeredgecolor='black',
    label='Recta de regresión'
)

# --- Título ---
plt.title(
    'Diagrama de Dispersión con Recta de Regresión',
    fontsize=14,
    fontweight='bold'
)

# --- Etiquetas de los ejes ---
plt.xlabel(
    'Promedio de exámenes cortos',
    fontsize=12
)

plt.ylabel(
    'Promedio final',
    fontsize=12
)

# --- Fuente de los ticks ---
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

# --- Márgenes ---
plt.margins(x=0.05, y=0.05)  # espacio extra alrededor de los datos
plt.gca().spines['right'].set_visible(False) # derecha
plt.gca().spines['top'].set_visible(False)   # superior

# Para eliminar márgenes completamente, usar:
# plt.margins(0)

# --- Cuadrícula (opcional, pero didáctica) ---
plt.grid(
    visible=True,
    linestyle='--',
    linewidth=0.7,
    alpha=0.1,
    color="gray"
)

# --- Leyenda ---
plt.legend(
    fontsize=10,
    loc='best',
    frameon=True
)

# --- Nota al pie ---
plt.text(
    0.4, -0.2,
    'Fuente: Datos del problema',
    fontsize=8,
    ha='left',
    va='center',
    transform=plt.gca().transAxes
)

# --- Guardar gráfico ---
plt.savefig(
    "grafico_dispersion_linea.png",
    bbox_inches='tight'
    )

plt.show()
```

3. ¿Los datos soportan la suposición de linealidad?
Sí, los datos soportan la suposición de linealidad

5. El 74.75% de la variación en las calificaciones finales

```python
from sklearn.metrics import r2_score

r2 = r2_score(y, y_calculada)
print(f'Coeficiente de determinación: {r2:.2%}\n')
```

6. Por cada punto que aumenta el promedio de los exámenes cortos, se espera que la calificación final aumente aproximadamente 0.64 puntos

```python
import statsmodels.api as sm
x_constante = sm.add_constant(x)
modelo = sm.OLS(y, x_constante).fit()
y_calculada = modelo.predict(x_constante)

modelo.params
```

$$
\hat {y} =24.526822+0.64180X
$$
donde
- $\hat{y}$es el promedio final calculado por el modelo
- $X$ es el promedio de examenes cortos

7. Con un 95% de confianza, la pendiente verdadera se encuentra entre 0.4192 y 0.8671

```python
# intervalo de confianza = 1 - alpha

modelo.conf_int(alpha = 0.05)
```

9. No se rechaza la hipótesis nula, asi que pueden considerarse aproximadamente normales

```python
from scipy.stats import shapiro
import matplotlib.pyplot as plt
import seaborn as sns

residuos = modelo.resid

# Prueba de Shapiro-Wilk
stat, valor_p_sh = shapiro(residuos)
print(f"valor-p (Shapiro) = {valor_p_sh}")

# Visualización: Q-Q plot
sm.qqplot(residuos, line='s')
plt.title("Q-Q plot de residuos")
plt.show()

# Histograma
sns.histplot(residuos, kde=True)
plt.title("Histograma de residuos")
plt.show()
```

4. El resultado muestra una correlación positiva, esto significa que los estudiantes que tienen mejores promedios en los exámenes cortos tienden a obtener también mejores calificaciones finales

```python
from scipy.stats import pearsonr

# Test de Pearson
# H0: rho = 0     (No hay correlación)
# H1: rho ≠ 0     (Sí hay correlación)
# alpha = 0.05

r, valor_p = pearsonr(x, y)

print(f"Coeficiente de correlación: {r: 0.4f}")
print(f'Valor p: {valor_p: 0.4f}')
```

8. Los residuales se encuentran tanto por arriba como por debajo de cero y no se observa una forma de curva clara

```python
import matplotlib.pyplot as plt

# --- Configuración general del gráfico ---
plt.figure(
    figsize=(6, 4),   # tamaño de la figura (ancho, alto) en pulgadas
    dpi=120           # resolución del gráfico
)

# --- Gráfico de dispersión ---
plt.scatter(
    x, residuos,
    marker="o",       # forma
    color='blue',     # color de los puntos
    edgecolor='black',    # borde de los puntos
    alpha=0.8,            # transparencia
    s=30,                 # tamaño de los puntos
    label='Gastos Generales (residuales)' # etiqueta para la leyenda
)

plt.axhline(
    y=0,     # Donde está la línea horizontal
    label="Recta de regresión", # Etiqueta
    linestyle = "--", # Estilo de línea
    color = "black",  # color
    linewidth=1.0,   # ancho de línea
    )

# --- Título ---
plt.title(
    'Grafico de dispersion residual',
    fontsize=14,
    fontweight='bold'
)

# --- Etiquetas de los ejes ---
plt.xlabel(
    'Examenes cortos',
    fontsize=12
)

plt.ylabel(
    'Gastos generales',
    fontsize=12
)

# --- Fuente de los ticks ---
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

# --- Márgenes ---
plt.margins(x=0.05, y=0.05)  # espacio extra alrededor de los datos
plt.gca().spines['right'].set_visible(False) # derecha
plt.gca().spines['top'].set_visible(False)   # superior

# Para eliminar márgenes completamente, usar:
# plt.margins(0)

# --- Cuadrícula (opcional, pero didáctica) ---
plt.grid(
    visible=True,
    linestyle='--',
    linewidth=0.7,
    alpha=0.1,
    color="gray"
)

# --- Leyenda ---
plt.legend(
    fontsize=10,
    loc='lower right', # best
    frameon=True
)

# --- Nota al pie ---
plt.text(
    0.4, -0.2,
    '',
    fontsize=8,
    ha='left',
    va='center',
    transform=plt.gca().transAxes
)

# --- Guardar gráfico ---
plt.savefig(
    "grafico_dispersion_residuales.png",
    bbox_inches='tight'
    )

plt.show()
```

10. no hay evidencia suficiente para decir que la variabilidad de los residuales cambie

```python
# Test de Breusch-Pagan
# H0: No hay heterocedasticidad (dispersión es igual en toda la recta)
# H1: Hay Heterocedasticidad (dispersión distinta a lo largo de la recta)
# alpha = 0.05
from statsmodels.stats.api import het_breuschpagan
_, valor_p_bp, _, _ = het_breuschpagan(residuos, x_constante)
print(f'valor_p de Breusch-Pagan: {valor_p_bp: 0.4f}\n')
```

12. Esto indica que existe una relación lineal significativa entre el promedio de los exámenes cortos y la calificación final

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
# Y ~ X
modelo_lineal = ols('promedio_final ~ promedio_de_examenes_cortos', data = df).fit()
tabla_anova = sm.stats.anova_lm(modelo_lineal)
tabla_anova
```

```python
import pandas as pd
link_csv= "https://raw.githubusercontent.com/2005c-h/EstadisticaInferencial2026A/refs/heads/main/Unidad01/Problemario/datos_problema_05.csv"
df=pd.read_csv(link_csv)
df
```

1. Se eligieron asi las variables ya que para saber la calificacion hay que saber las interrupciones que tiene

```python
x = df['número_interrupciones_al_trabajador']
y = df['calificación_del_trabajador_en_la_prueba_de_hostilidad']
```

2. Al observar los puntos, se nota que tienen una dirección descendente bastante marcada

```python
import matplotlib.pyplot as plt

# --- Configuración general del gráfico ---
plt.figure(
    figsize=(6, 4),   # tamaño de la figura (ancho, alto) en pulgadas
    dpi=120           # resolución del gráfico
)

# --- Gráfico de dispersión ---
plt.scatter(
    x, y,
    marker='*',       # forma
    color='red',     # color de los puntos
    edgecolor='red',    # borde de los puntos
    alpha=0.9,            # transparencia
    s=50,                 # tamaño de los puntos
    label='Calificacion del trabajador en l prueba de hostilidad' # etiqueta para la leyenda
)

# --- Título ---
plt.title(
    'Gráfico de dispersión de interrupciones y calificaciones',
    fontsize=14,
    fontweight='bold'
)

# --- Etiquetas de los ejes ---
plt.xlabel(
    'Numerode interrupciones',
    fontsize=12
)

plt.ylabel(
    'Calificaciones de los trabajadores',
    fontsize=12
)

# --- Fuente de los ticks ---
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

# --- Márgenes ---
plt.margins(x=0.05, y=0.05)  # espacio extra alrededor de los datos
plt.gca().spines['right'].set_visible(False) # derecha
plt.gca().spines['top'].set_visible(False)   # superior

# --- Cuadrícula (opcional, pero didáctica) ---
plt.grid(
    visible=True,
    linestyle='--',
    linewidth=0.7,
    alpha=0.1,
    color="gray"
)

# --- Leyenda ---
plt.legend(
    fontsize=10,
    loc='best',
    frameon=True
)

# --- Nota al pie ---
plt.text(
    0.4, -0.2,
    '',
    fontsize=8,
    ha='left',
    va='center',
    transform=plt.gca().transAxes
)

# --- Guardar gráfico ---
plt.savefig(
    "grafico_dispersion.png",
    bbox_inches='tight'
    )

plt.show()
```

3. ¿Los datos soportan la suposición de linealidad?

si, ya que los puntos se encuentran bastante cerca de una línea recta descendente

4. El valor obtenido está muy cerca de -1, por lo que hay una relación negativa, mientras más veces se interrumpe al trabajador, menor tiende a ser su calificación en la prueba

```python
from scipy.stats import pearsonr

# Test de Pearson
# H0: rho = 0     (No hay correlación)
# H1: rho ≠ 0     (Sí hay correlación)
# alpha = 0.05

r, valor_p = pearsonr(x, y)

print(f"Coeficiente de correlación: {r: 0.4f}")
print(f'Valor p: {valor_p: 0.4f}')
```

```python
import statsmodels.api as sm
x_constante = sm.add_constant(x)
modelo = sm.OLS(y, x_constante).fit()
y_calculada = modelo.predict(x_constante)

modelo.params
```

$$
\hat {y} =70.5-2.8X
$$
donde
- $\hat{y}$es la calificacion del trabajador  
- $X$ es el numero de interrupciones del trabajador

6. Por cada interrupción adicional al trabajador, la calificación esperada en la prueba disminuye aproximadamente 2.8 puntos

5. El modelo logra explicar aproximadamente el 98.58% de los cambios que se presentan en las calificaciones de la prueba

```python
from sklearn.metrics import r2_score  # recomendada
r2 = r2_score(y, y_calculada)
print(f'Coeficiente de determinación: {r2: 0.2%}\n')
```

7. Con un 95% de confianza, la disminución de la calificación por cada interrupción se encuentra entre aproximadamente 2.46 y 3.14 puntos

```python
# intervalo de confianza = 1 - alpha

modelo.conf_int(alpha = 0.05)
```

9. El valor de p es un poco mayor que 0.05, por lo que no se rechaza la normalidad de los residuales

```python
from scipy.stats import shapiro
import matplotlib.pyplot as plt
import seaborn as sns

residuos = modelo.resid

# Prueba de Shapiro-Wilk
stat, valor_p_sh = shapiro(residuos)
print(f"valor-p (Shapiro) = {valor_p_sh}")

# Visualización: Q-Q plot
sm.qqplot(residuos, line='s')
plt.title("Q-Q plot de residuos")
plt.show()

# Histograma
sns.histplot(residuos, kde=True)
plt.title("Histograma de residuos")
plt.show()
```

10. El resultado no muestra evidencia de que la variación de los errores cambie conforme aumentan las interrupciones

```python
# Test de Breusch-Pagan
# H0: No hay heterocedasticidad (dispersión es igual en toda la recta)
# H1: Hay Heterocedasticidad (dispersión distinta a lo largo de la recta)
# alpha = 0.05

from statsmodels.stats.api import het_breuschpagan
_, valor_p_bp, _, _ = het_breuschpagan(residuos, x_constante)
print(f'valor_p de Breusch-Pagan: {valor_p_bp: 0.4f}\n')
```

11. 11. Interpolación y extrapolación

La ecuación que utilizaremos es:

$$ \hat Y=70.5-2.8X $$

Los valores originales de X van de 5 a 25 interrupciones.

Interpolación 1: 12 interrupciones


$$ \hat Y=70.5-2.8(12) $$ $$ \hat Y=70.5-33.6 $$ $$ \boxed{\hat Y=36.9} $$


Para un trabajador que recibe 12 interrupciones, el modelo estima una calificación de aproximadamente 36.9 puntos.

Interpolación 2: 18 interrupciones


$$ \hat Y=70.5-2.8(18) $$ $$ \boxed{\hat Y=20.1} $$


Con 18 interrupciones, la calificación esperada sería aproximadamente de 20.1 puntos.

Extrapolación: 30 interrupciones


$$ \hat Y=70.5-2.8(30) $$ $$ \boxed{\hat Y=-13.5} $$


Para 30 interrupciones, la ecuación daría una calificación de -13.5 puntos

12. Esto indica que la relación entre el número de interrupciones y la calificación de la prueba es estadísticamente significativa

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
# Y ~ X
modelo_lineal = ols('calificación_del_trabajador_en_la_prueba_de_hostilidad  ~ número_interrupciones_al_trabajador', data = df).fit()
tabla_anova = sm.stats.anova_lm(modelo_lineal)
tabla_anova
```
